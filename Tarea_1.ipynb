# Transformada Discreta de Fourier No Uniforme (NUDFT)

## ¿Cuándo usar la NUDFT?

Se emplea la NUDFT cuando:
1. **El muestreo temporal es irregular** (las muestras no están equiespaciadas en el tiempo)
2. **Se requiere el espectro en frecuencias uniformes** (como en una DFT estándar)

Aplicaciones típicas incluyen:
- Procesamiento de datos experimentales con intervalos variables
- Análisis de señales naturales (observaciones astronómicas, mediciones geofísicas)
- Casos donde el muestreo uniforme no es posible o práctico

## Definición de NUDFT

La NUDFT generaliza la DFT estándar para manejar muestreo no uniforme. Calcula componentes espectrales en frecuencias equiespaciadas ωₙ a partir de muestras en tiempos arbitrarios tₖ:

$$
X(\omega_n) = \sum_{k=0}^{N-1} x(t_k) e^{-i \omega_n t_k}
$$

donde:
- tₖ = tiempos de muestreo no uniformes
- ωₙ = 2πn/N (frecuencias uniformes, iguales a la DFT estándar)
- x(tₖ) = valores de la señal en los tiempos tₖ

## Interpretación Física

La NUDFT:
1. Proyecta la señal irregular sobre una base de exponenciales complejas
2. Preserva la relación tiempo-frecuencia de la DFT estándar
3. Evita artefactos de interpolación forzada
4. Mantiene la interpretación espectral tradicional

**Nota importante**: La NUDFT es crucial cuando el muestreo irregular no puede ignorarse pero se necesita un análisis espectral comparable al de una FFT tradicional.
