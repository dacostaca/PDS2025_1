```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import convolve
from scipy.fft import fft, fftfreq, fftshift
from ipywidgets import interact, FloatSlider

# Parámetros de la señal
fs = 1000        # Frecuencia de muestreo (Hz)
T = 1            # Duración (s)
t = np.linspace(0, T, int(fs*T), endpoint=False)

# Señal compuesta (4 componentes)
frecuencias_base = [50, 100, 150, 200]
senal_compuesta = (
    np.sin(2 * np.pi * frecuencias_base[0] * t) +
    np.cos(2 * np.pi * frecuencias_base[1] * t) +
    np.sin(2 * np.pi * frecuencias_base[2] * t) +
    np.cos(2 * np.pi * frecuencias_base[3] * t)
)

# Función para calcular la FFT
def calcular_fft(signal):
    N = len(signal)
    spectrum = np.abs(fftshift(fft(signal)))
    freqs = fftshift(fftfreq(N, 1/fs))
    return freqs, spectrum

# Función interactiva
def actualizar(frecuencia_variable):
    senal_variable = np.sin(2 * np.pi * frecuencia_variable * t)
    resultado_convolucion = convolve(senal_compuesta, senal_variable, mode='same')

    # FFTs
    f1, sp1 = calcular_fft(senal_compuesta)
    f2, sp2 = calcular_fft(senal_variable)
    f3, sp3 = calcular_fft(resultado_convolucion)

    # Gráficas
    plt.figure(figsize=(14, 8))

    plt.subplot(3, 1, 1)
    plt.plot(f1, sp1)
    plt.title("Espectro de la señal compuesta")
    plt.grid(True)

    plt.subplot(3, 1, 2)
    plt.plot(f2, sp2)
    plt.title(f"Espectro de la señal variable (f = {frecuencia_variable:.1f} Hz)")
    plt.grid(True)

    plt.subplot(3, 1, 3)
    plt.plot(f3, sp3)
    plt.title("Espectro de la convolución")
    plt.grid(True)

    plt.tight_layout()
    plt.show()

# Interfaz interactiva
interact(
    actualizar,
    frecuencia_variable=FloatSlider(
        value=10,
        min=10,
        max=300,
        step=1,
        description='Frecuencia (Hz)',
        continuous_update=False
    )
)
```
